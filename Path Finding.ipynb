```python
import tkinter as tk
from tkinter import *
from tkinter import ttk


CELL_SIZE = 20
grid = [[Tile() for i in range(50)]for i in range(50)]

class Tile:
    def __init__(self):
        self.color = None
        self.id = None

def draw(canvas):
    y = 0
    x = 0
    for row in grid:    
        for tile in row:
            tile.id = canvas.create_rectangle(x, y, x + CELL_SIZE, y + CELL_SIZE)
            y += CELL_SIZE
        x += CELL_SIZE
        y = 0
    canvas.pack()
    canvas.bind('<Button-1>', changeColor)
    canvas.bind("<B1-Motion>", handleMouseMotion)
    canvas.bind("<ButtonRelease-1>", switchColor)

def eventCoords(event):
    row = int(event.x / CELL_SIZE)
    column = int(event.y / CELL_SIZE)
    return row, column

def changeColor(event):
    row, column = eventCoords(event)
#     print("Row : ",row)
#     print("Column : ",column)
#     print("EVENT : ",event)
    if grid[row][column].color == 'red':
        canvas.itemconfigure(grid[row][column].id, fill = 'white')
        grid[row][column].color = None
    else:
        canvas.itemconfigure(grid[row][column].id, fill = 'red')
        grid[row][column].color = 'red'

hovered = []    
def handleMouseMotion(event):
    row, column = eventCoords(event)
    
    if grid[row][column].id not in hovered:
#         if grid[row][column].color == 'red':
        canvas.itemconfigure(grid[row][column].id, fill = 'black')
        grid[row][column].color = 'black'
        hovered.append(grid[row][column].id)
    else:
        canvas.itemconfigure(grid[row][column].id, fill = 'white')
        grid[row][column].color = None
        hovered.remove(grid[row][column].id)

def switchColor():
#     hovered.clear()
    return None
    
master = Tk()
canvas = Canvas(master, width = CELL_SIZE * 50, height = CELL_SIZE * 50)
btn = Button(master, text = 'Start Node', bd = '5')
btn.pack(side = 'left')

#Drop down creation...
n = tk.StringVar()
pathAlgo = ttk.Combobox(master, width = 27, textvariable = n)
pathAlgo['values'] = ('A*',
                     'Dijiktras',
                     'BFS',
                     'DFS',
                     'Tree')
# pathAlgo.grid(column = 1, row = 5)
pathAlgo.current()
canvas.pack()
# grid
draw(canvas)
master.mainloop()
```
